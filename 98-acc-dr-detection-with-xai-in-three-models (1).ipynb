# ***Components of CNN***

The CNN model works in two steps: feature extraction and Classification[1]

Feature Extraction is a phase where various filters and layers are applied to the images to extract the information and features out of it and once it’s done it is passed on to the next phase i.e Classification where they are classified based on the target variable of the problem.

A typical CNN model looks like this:

    Input layer
    Convolution layer + Activation function
    Pooling layer
    Fully Connected Layer

![](https://editor.analyticsvidhya.com/uploads/75211cnn-schema1.jpg)


Let’s learn about each layer in detail.

### Input layer

As the name says, it’s our input image and can be Grayscale or RGB. Every image is made up of pixels that range from 0 to 255. We need to normalize them i.e convert the range between 0 to 1  before passing it to the model.

Below is the example of an input image of size 4*4 and has 3 channels i.e RGB and pixel values.


![](https://editor.analyticsvidhya.com/uploads/98920pixel%20image.png)



### Convolution Layer

The convolution layer is the layer where the filter is applied to our input image to extract or detect its features. A filter is applied to the image multiple times and creates a feature map which helps in classifying the input image. Let’s understand this with the help of an example. For simplicity, we will take a 2D input image with normalized pixels.

![](https://editor.analyticsvidhya.com/uploads/57527Screenshot%20(155).png)

In the above figure, we have an input image of size 6*6 and applied a filter of 3*3 on it to detect some features. In this example, we have applied only one filter but in practice, many such filters are applied to extract information from the image.

The result of applying the filter to the image is that we get a Feature Map of 4*4 which has some information about the input image. Many such feature maps are generated in practical applications.

Let’s get into some maths behind getting the feature map in the above image.

![](https://editor.analyticsvidhya.com/uploads/89436Screenshot%20(156).png)



As presented in the above figure, in the first step the filter is applied to the green highlighted part of the image, and the pixel values of the image are multiplied with the values of the filter (as shown in the figure using lines) and then summed up to get the final value.

In the next step, the filter is shifted by one column as shown in the below figure. This jump to the next column or row is known as stride and in this example, we are taking a stride of 1 which means we are shifting by one column.

![](https://editor.analyticsvidhya.com/uploads/47821Screenshot%20(158).png)


Similarly, the filter passes over the entire image and we get our final Feature Map. Once we get the feature map, an activation function is applied to it for introducing nonlinearity.

A point to note here is that the Feature map we get is smaller than the size of our image. As we increase the value of stride the size of the feature map decreases.

![](https://editor.analyticsvidhya.com/uploads/86290Screenshot%20(159).png)

This is how a filter passes through the entire image with the stride of 1

### Pooling Layer

The pooling layer is applied after the Convolutional layer and is used to reduce the dimensions of the feature map which helps in preserving the important information or features of the input image and reduces the computation time.

Using pooling, a lower resolution version of input is created that still contains the large or important elements of the input image.

The most common types of Pooling are Max Pooling and Average Pooling. The below figure shows how Max Pooling works. Using the Feature map which we got from the above example to apply Pooling. Here we are using a Pooling layer of size 2*2 with a stride of 2.

The maximum value from each highlighted area is taken and a new version of the input image is obtained which is of size 2*2 so after applying Pooling the dimension of the feature map has reduced. 

![](https://editor.analyticsvidhya.com/uploads/30447Screenshot%20(160).png)


### Fully Connected Layer

Till now we have performed the Feature Extraction steps, now comes the Classification part. The Fully connected layer (as we have in ANN) is used for classifying the input image into a label. This layer connects the information extracted from the previous steps (i.e Convolution layer and Pooling layers) to the output layer and eventually classifies the input into the desired label.

Progression to vision impairment can be slowed or averted if DR is detected in time, however this can be difficult as the disease often shows few symptoms until it is too late to provide effective treatment.

Currently, detecting DR is a time-consuming and manual process that requires a trained clinician to examine and evaluate digital color fundus photographs of the retina. Clinicians can identify DR by the presence of lesions associated with the vascular abnormalities caused by the disease. While this approach is effective, its resource demands are high. The expertise and equipment required are often lacking in areas where the rate of diabetes in local populations is high and DR detection is most needed. As the number of individuals with diabetes continues to grow, the infrastructure needed to prevent blindness due to DR will become even more insufficient.

### ****Understanding the Stages of Diabetic Retinopathy****

Elevated blood sugar, blood pressure and cholesterol levels and increased body weight are associated with uncontrolled diabetes and can damage the delicate blood vessels of the retina, causing a disease called diabetic retinopathy. In the early stages of diabetic retinopathy, vision loss may be prevented or limited; but as the condition advances, it becomes more difficult to prevent vision loss.

![](https://www.elmanretina.com/wp-content/uploads/shutterstock_657947524-1024x700.jpg)

****Non-Proliferative Retinopathy****

Non-proliferative diabetic retinopathy is the earlier and less serious stage of the disease. Due to damage from elevated blood sugar levels, the tiny blood vessels of the retina start to swell and leak fluid and blood. The leaking causes swelling inside the retina called macular edema, which is a common cause of visual impairment in people with diabetes.

As non-proliferative retinopathy advances, the blood vessels may become blocked or closed off, and new abnormal blood vessels may start to grow and form in the wrong parts of the retina. Vision is usually unaffected in the non-proliferative stages, but it is best for the retina doctor to watch these changes closely every few months to ensure it doesn’t worsen to the “proliferative” stage.

****Proliferative Diabetic Retinopathy****

The more serious form of the disease, proliferative diabetic retinopathy is characterized by a process known as neovascularization. When the blood vessels of the retina become blocked, the retina is deprived of the oxygen and nutrients it needs to thrive. Sensing its limited blood supply, the retina responds by growing new abnormal blood vessels in the wrong parts of the retina. But the new blood vessels are abnormal and do not provide proper blood flow. They are extremely fragile and, as they grow, can leak blood into the vitreous. Sometimes this bleeding causes visual distortions called floaters; if the bleeding is severe enough, it can partially or completely impair vision.

In some cases, the formation of the new, abnormal blood vessels leads to the development of scar tissue, or fibrosis, in the retina. Scar tissue may cause the retina to wrinkle or “pucker,” or even detach from its normal position along the back of the eye, which is referred to as “tractional retinal detachment.” Depending on the complication, central vision loss can occur.

Another complication to watch out for is the growth of these abnormal blood vessels blocking the drainage angle, the natural mechanism by which fluid exits the eye. If this occurs, pressure inside the eye can spike very quickly, causing neovascular angle-closure glaucoma.

### Data Preparation and Augmentation

In this step, we import necessary libraries and configure data augmentation pipelines using `ImageDataGenerator`.  
Augmentation techniques such as rotation, zoom, shear, and horizontal flipping are applied to enhance model generalization.  
The input images are resized to 300x300 pixels to match the input size expected by EfficientNetB3.  
Separate generators are created for training, validation, and testing datasets, with preprocessing aligned to the EfficientNet requirements.


```python
from tensorflow.keras.applications import EfficientNetB3
from tensorflow.keras.applications.efficientnet import preprocess_input
from tensorflow.keras.models import Model
from tensorflow.keras import layers, optimizers
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
import os
from tensorflow.keras.metrics import AUC

def create_data_generators(train_path, val_path, test_path, preprocess_func, target_size, batch_size=32):
    train_gen = ImageDataGenerator(
        preprocessing_function=preprocess_func,
        rotation_range=15,
        zoom_range=0.2,
        shear_range=0.2,
        horizontal_flip=True,
        width_shift_range=0.1,
        height_shift_range=0.1
    )
    val_gen = ImageDataGenerator(preprocessing_function=preprocess_func)
    test_gen = ImageDataGenerator(preprocessing_function=preprocess_func)

    train_batches = train_gen.flow_from_directory(train_path, target_size=target_size, batch_size=batch_size, class_mode='binary')
    val_batches = val_gen.flow_from_directory(val_path, target_size=target_size, batch_size=batch_size, class_mode='binary')
    test_batches = test_gen.flow_from_directory(test_path, target_size=target_size, batch_size=batch_size, class_mode='binary', shuffle=False)

    return train_batches, val_batches, test_batches
```

```python
def set_seed(seed=42):
    np.random.seed(seed)
    random.seed(seed)
    tf.random.set_seed(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)

```

# ****1. EfficientNetB3 Model****

```python
set_seed(42)
# Build Model
train_batches, val_batches, test_batches = create_data_generators(
    train_path='train',
    val_path='val',
    test_path='test',
    preprocess_func=preprocess_input,
    target_size=(300,300),
    batch_size=32
)
base_model = EfficientNetB3(include_top=False, input_shape=(300,300,3), pooling='avg', weights='imagenet')
base_model.trainable = False

x = base_model.output
x = layers.Dense(128, activation='relu')(x)
x = layers.Dropout(0.4)(x)
output = layers.Dense(1, activation='sigmoid')(x)

model = Model(inputs=base_model.input, outputs=output)

# Compile
model.compile(
    optimizer=optimizers.Adam(learning_rate=1e-4),
    loss='binary_crossentropy',
    metrics=['accuracy',AUC(name='auc')]
)

# Callbacks
early_stop = EarlyStopping(patience=3, restore_best_weights=True)
lr_scheduler = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=2, verbose=1)

# Initial training
history = model.fit(
    train_batches,
    epochs=10,
    validation_data=val_batches,
    callbacks=[early_stop, lr_scheduler]
)

```

```python
set_seed(42)
# Fine-Tuning
base_model.trainable = True

for layer in base_model.layers[:50]:
    layer.trainable = False

model.compile(
    optimizer=optimizers.Adam(1e-5),
    loss='binary_crossentropy',
    metrics=['accuracy' , AUC(name='auc')]
)

history_finetune = model.fit(
    train_batches,
    epochs=10,
    validation_data=val_batches,
    callbacks=[early_stop, lr_scheduler]
)
```

```python
from sklearn.metrics import roc_curve, auc

y_pred_prob = model.predict(test_batches).ravel()
y_true = test_batches.classes

fpr, tpr, thresholds = roc_curve(y_true, y_pred_prob)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(8,6))
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (AUC = {roc_auc:.3f})')
plt.plot([0,1], [0,1], color='navy', lw=2, linestyle='--')
plt.xlim([0.0,1.0])
plt.ylim([0.0,1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic (ROC)')
plt.legend(loc='lower right')
plt.show()
```

```python

# Loss plot
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)  # 1 row, 2 columns, first plot
plt.plot(history_finetune.history['loss'], label='Training Loss')
plt.plot(history_finetune.history['val_loss'], label='Validation Loss')
plt.title('Loss Over Epochs (Fine-Tuning)')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

# Accuracy plot
plt.subplot(1, 2, 2)  # 1 row, 2 columns, second plot
plt.plot(history_finetune.history['accuracy'], label='Training Accuracy')
plt.plot(history_finetune.history['val_accuracy'], label='Validation Accuracy')
plt.title('Accuracy Over Epochs (Fine-Tuning)')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

**Model Evaluation at Varying Thresholds**

```python
from sklearn.metrics import confusion_matrix ,classification_report

# Ensure the test_batches are reset before prediction
test_batches.reset()
predictions_proba = model.predict(test_batches)

# True labels
y_true = test_batches.classes

# Class names: Ensure index 0 is 'DR' and 1 is 'No_DR'
class_names = ['DR', 'No_DR']

print("\n--- Model Evaluation at Different Thresholds ---")

# Try a range of thresholds from 0.10 to 0.95
thresholds = np.arange(0.10, 1.0, 0.05)

# Separate best values for each class
best_f1_nodr = -1
best_threshold_nodr = -1
best_f1_dr = -1
best_threshold_dr = -1

for threshold in thresholds:
    y_pred_binary = (predictions_proba > threshold).astype(int)

    # Generate classification report
    report = classification_report(y_true, y_pred_binary, target_names=class_names, output_dict=True)

    print(f"\n--- Threshold: {threshold:.2f} ---")
    print(f"Accuracy: {report['accuracy']:.4f}")
    print(f"  DR     - Precision: {report['DR']['precision']:.4f}, Recall: {report['DR']['recall']:.4f}, F1-score: {report['DR']['f1-score']:.4f}")
    print(f"  No_DR  - Precision: {report['No_DR']['precision']:.4f}, Recall: {report['No_DR']['recall']:.4f}, F1-score: {report['No_DR']['f1-score']:.4f}")

    # Best for No_DR
    if report['No_DR']['f1-score'] > best_f1_nodr:
        best_f1_nodr = report['No_DR']['f1-score']
        best_threshold_nodr = threshold

    # Best for DR
    if report['DR']['f1-score'] > best_f1_dr:
        best_f1_dr = report['DR']['f1-score']
        best_threshold_dr = threshold

print(f"\n--- Best Threshold (based on F1-score for No_DR): {best_threshold_nodr:.2f} (F1-score: {best_f1_nodr:.4f}) ---")
print(f"--- Best Threshold (based on F1-score for DR): {best_threshold_dr:.2f} (F1-score: {best_f1_dr:.4f}) ---")

# Confusion matrix for No_DR's best threshold
best_cm = confusion_matrix(y_true, (predictions_proba > best_threshold_nodr).astype(int))
plt.figure(figsize=(8,6))
sns.heatmap(best_cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=class_names, yticklabels=class_names)
plt.title(f'Best Threshold Confusion Matrix for No_DR ({best_threshold_nodr:.2f})')
plt.show()
```

### Why these validation results are not sufficient for final model selection:

- Validation results can be influenced by model selection and hyperparameter tuning during training, potentially leading to slight **overfitting on the validation set**.
- Therefore, it is essential to evaluate the models on an **independent test set** that was not used during training or validation.
- Testing on an unseen dataset ensures an unbiased and objective assessment of the model’s generalization ability in real-world scenarios.
- Without this final evaluation, selecting a model solely based on validation performance may lead to biased conclusions and inaccurate estimates of true performance.

```python
from tensorflow.keras.applications.efficientnet import preprocess_input as efficientnet_preprocess_input


def create_test_generator(test_path, preprocess_func, target_size, batch_size=32):
    from tensorflow.keras.preprocessing.image import ImageDataGenerator
    test_gen = ImageDataGenerator(preprocessing_function=preprocess_func)
    test_batches = test_gen.flow_from_directory(
        test_path,
        target_size=target_size,
        batch_size=batch_size,
        class_mode='binary',
        shuffle=False
    )
    return test_batches
test_path = "test" 
# استدعاء معالجات النماذج (تأكد من وجود test_path صحيح)
test_batches_resnet = create_test_generator(test_path, resnet_preprocess_input, (224,224))
test_batches_densenet = create_test_generator(test_path, densenet_preprocess_input, (224,224))
test_batches_efficientnet = create_test_generator(test_path, efficientnet_preprocess_input, (300,300))

models = {
    'ResNet50V2': ('ResNet50V2.model', test_batches_resnet),
    'DenseNet201': ('DenseNet201.model', test_batches_densenet),
    'EfficientNetB3': ('EfficientNetB3.model', test_batches_efficientnet)
}

plt.figure(figsize=(9,7))
plt.plot([0,1], [0,1], linestyle='--', color='gray', lw=1, label='Random Guess')

colors = ['#1f77b4', '#ff7f0e', '#2ca02c']

for i, (name, (model_path, test_batches)) in enumerate(models.items()):
    model = load_model(model_path, compile=True)
    y_true = test_batches.classes
    y_pred_prob = model.predict(test_batches).ravel()
    fpr, tpr, _ = roc_curve(y_true, y_pred_prob)
    roc_auc = auc(fpr, tpr)
    
    plt.plot(fpr, tpr, color=colors[i], lw=2, label=f'{name} (AUC = {roc_auc:.3f})')
    
    loss, accuracy, auc_metric = model.evaluate(test_batches, verbose=0)
    print(f'{name}: Loss={loss:.4f}, Accuracy={accuracy:.4f}, AUC={auc_metric:.4f}')

plt.xlabel('False Positive Rate', fontsize=13)
plt.ylabel('True Positive Rate', fontsize=13)
plt.title('ROC Curve Comparison', fontsize=16, fontweight='bold')
plt.legend(loc='lower right', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

## ***Generate Grad-CAM heatmap for a given image and model***

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.applications.resnet_v2 import preprocess_input as resnet_preprocess_input
from tensorflow.keras.applications.densenet import preprocess_input as densenet_preprocess_input
from tensorflow.keras.applications.efficientnet import preprocess_input as efficientnet_preprocess_input
import os

models = {
    'ResNet50V2': {
        'model': tf.keras.models.load_model('ResNet50V2.model'),
        'preprocess': resnet_preprocess_input,
        'last_conv_layer': 'conv5_block3_3_conv', 
        'input_size': (224, 224)
    },
    'DenseNet201': {
        'model': tf.keras.models.load_model('DenseNet201.model'),
        'preprocess': densenet_preprocess_input,
        'last_conv_layer': 'conv5_block32_2_conv', 
        'input_size': (224, 224)
    },
    'EfficientNetB3': {
        'model': tf.keras.models.load_model('EfficientNetB3.model'),
        'preprocess': efficientnet_preprocess_input,
        'last_conv_layer': 'top_conv', 
        'input_size': (300, 300)
    }
}

def make_gradcam_heatmap(img_array, model, last_conv_layer_name, pred_index=None):
    grad_model = tf.keras.models.Model(
        [model.inputs], 
        [model.get_layer(last_conv_layer_name).output, model.output]
    )
    
    with tf.GradientTape() as tape:
        conv_outputs, predictions = grad_model(img_array)
        if pred_index is None:
            pred_index = tf.argmax(predictions[0])
        class_channel = predictions[:, pred_index]
    
    grads = tape.gradient(class_channel, conv_outputs)
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))
    conv_outputs = conv_outputs[0]
    heatmap = conv_outputs @ pooled_grads[..., tf.newaxis]
    heatmap = tf.squeeze(heatmap)
    heatmap = tf.maximum(heatmap, 0) / tf.math.reduce_max(heatmap)
    return heatmap.numpy()

def get_prediction_and_heatmap(model_info, img_path):
    model = model_info['model']
    preprocess_func = model_info['preprocess']
    last_conv_layer = model_info['last_conv_layer']
    target_size = model_info['input_size']
    
    img = cv2.imread(img_path)
    if img is None:
        raise FileNotFoundError(f"Image not found: {img_path}")
    
    img_resized = cv2.resize(img, target_size)
    img_rgb = cv2.cvtColor(img_resized, cv2.COLOR_BGR2RGB)
    
    img_preprocessed = preprocess_func(img_rgb.astype('float32'))
    img_input = np.expand_dims(img_preprocessed, axis=0)
    
    pred = model.predict(img_input)[0][0]
    predicted_class = "No DR" if pred > 0.5 else "DR"
    
    heatmap = make_gradcam_heatmap(img_input, model, last_conv_layer)
    
    return img_rgb, predicted_class, pred, heatmap

def plot_all_models_for_images(image_paths):
    n_images = len(image_paths)
    n_models = len(models)
    
    fig, axes = plt.subplots(2, n_models * n_images, figsize=(4 * n_models * n_images, 8))
    
    for i, img_path in enumerate(image_paths):
        for j, (model_name, model_info) in enumerate(models.items()):
            img_rgb, predicted_class, prob, heatmap = get_prediction_and_heatmap(model_info, img_path)
            
            # صف الصور مع النتائج
            ax_img = axes[0, i * n_models + j]
            ax_img.imshow(img_rgb)
            ax_img.axis('off')
            ax_img.set_title(f"{os.path.basename(img_path)}\n{model_name}\nPrediction: {predicted_class}\nProb: {prob:.2f}", fontsize=10)
            
            # صف خرائط الحرارة
            ax_heat = axes[1, i * n_models + j]
            heatmap_resized = cv2.resize(heatmap, (img_rgb.shape[1], img_rgb.shape[0]))
            heatmap_resized = np.uint8(255 * heatmap_resized)
            heatmap_color = cv2.applyColorMap(heatmap_resized, cv2.COLORMAP_JET)
            superimposed_img = cv2.addWeighted(img_rgb, 0.6, heatmap_color, 0.4, 0)
            ax_heat.imshow(superimposed_img)
            ax_heat.axis('off')
            ax_heat.set_title(f"{model_name} Grad-CAM", fontsize=10)
    
    plt.tight_layout()
    plt.show()

image_files = [
    "../input/diabetic-retinopathy-224x224-gaussian-filtered/gaussian_filtered_images/gaussian_filtered_images/Moderate/000c1434d8d7.png"

]
plot_all_models_for_images(image_files)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
import os

def make_scorecam_heatmap(img_array, model, last_conv_layer_name, preprocess_func, target_size, pred_index=None):
    conv_layer = model.get_layer(last_conv_layer_name)
    score_model = tf.keras.models.Model(
        [model.inputs], 
        [conv_layer.output, model.output]
    )
    
    conv_outputs, predictions = score_model(img_array)
    if pred_index is None:
        pred_index = tf.argmax(predictions[0])
    
    conv_outputs = conv_outputs[0]  # (H, W, N_features)
    num_features = conv_outputs.shape[-1]
    
    heatmap = np.zeros(conv_outputs.shape[:2], dtype=np.float32)
    
    for i in range(num_features):
        fmap = conv_outputs[..., i]
        
        fmap_min, fmap_max = tf.reduce_min(fmap), tf.reduce_max(fmap)
        if fmap_max - fmap_min == 0:
            continue
        norm_fmap = (fmap - fmap_min) / (fmap_max - fmap_min)
        
        upsampled_fmap = tf.image.resize(norm_fmap[..., tf.newaxis], target_size)
        
        masked_img = img_array[0] * upsampled_fmap.numpy()
        
        masked_input = np.expand_dims(masked_img, axis=0)
        score = model.predict(masked_input, verbose=0)[0][pred_index]
        
        heatmap += score * norm_fmap.numpy()
    
    heatmap = np.maximum(heatmap, 0)
    heatmap /= np.max(heatmap) if np.max(heatmap) != 0 else 1
    return heatmap

def get_prediction_and_scorecam(model_info, img_path):
    model = model_info['model']
    preprocess_func = model_info['preprocess']
    last_conv_layer = model_info['last_conv_layer']
    target_size = model_info['input_size']
    
    img = cv2.imread(img_path)
    if img is None:
        raise FileNotFoundError(f"Image not found: {img_path}")
    
    img_resized = cv2.resize(img, target_size)
    img_rgb = cv2.cvtColor(img_resized, cv2.COLOR_BGR2RGB)
    
    img_preprocessed = preprocess_func(img_rgb.astype('float32'))
    img_input = np.expand_dims(img_preprocessed, axis=0)
    
    pred = model.predict(img_input, verbose=0)[0][0]
    predicted_class = "No DR" if pred > 0.5 else "DR"
    
    heatmap = make_scorecam_heatmap(img_input, model, last_conv_layer, preprocess_func, target_size)
    
    return img_rgb, predicted_class, pred, heatmap

def plot_all_models_scorecam(image_paths):
    n_images = len(image_paths)
    n_models = len(models)
    
    fig, axes = plt.subplots(2, n_models * n_images, figsize=(4 * n_models * n_images, 8))
    
    for i, img_path in enumerate(image_paths):
        for j, (model_name, model_info) in enumerate(models.items()):
            img_rgb, predicted_class, prob, heatmap = get_prediction_and_scorecam(model_info, img_path)
            
            ax_img = axes[0, i * n_models + j]
            ax_img.imshow(img_rgb)
            ax_img.axis('off')
            ax_img.set_title(f"{os.path.basename(img_path)}\n{model_name}\nPrediction: {predicted_class}\nProb: {prob:.2f}", fontsize=10)
            
            heatmap_resized = cv2.resize(heatmap, (img_rgb.shape[1], img_rgb.shape[0]))
            heatmap_resized = np.uint8(255 * heatmap_resized)
            heatmap_color = cv2.applyColorMap(heatmap_resized, cv2.COLORMAP_JET)
            superimposed_img = cv2.addWeighted(img_rgb, 0.6, heatmap_color, 0.4, 0)
            
            ax_heat = axes[1, i * n_models + j]
            ax_heat.imshow(superimposed_img)
            ax_heat.axis('off')
            ax_heat.set_title(f"{model_name} Score-CAM", fontsize=10)
    
    plt.tight_layout()
    plt.show()

image_files = [
    "../input/diabetic-retinopathy-224x224-gaussian-filtered/gaussian_filtered_images/gaussian_filtered_images/Moderate/000c1434d8d7.png"
]
plot_all_models_scorecam(image_files)
```

### Preprocess APTOS image for evaluation: read, resize to target size, and convert to RGB.
Preprocesses and organizes new test images into class folders for evaluation and creates a DataFrame for data loading.










```python
import os
import shutil
import cv2
import pandas as pd
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications.resnet_v2 import preprocess_input as resnet_preprocess_input
from tensorflow.keras.applications.densenet import preprocess_input as densenet_preprocess_input
from tensorflow.keras.applications.efficientnet import preprocess_input as efficientnet_preprocess_input

# 1. تعريف دالة المعالجة للصور قبل التقييم
def preprocess_aptos_image_for_eval(image_base_dir, target_size=(300, 300)):
    img = cv2.imread(image_path)
    if img is None:
        return None
    img_resized = cv2.resize(img, target_size)
    if len(img_resized.shape) == 2:
        img_resized = cv2.cvtColor(img_resized, cv2.COLOR_GRAY2RGB)
    elif img_resized.shape[2] == 4:
        img_resized = cv2.cvtColor(img_resized, cv2.COLOR_BGRA2BGR)
    return img_resized

# 2. تحضير مجلد الصور المعالجة ومساراتها
processed_test_images_dir = 'processed_new_test_images_for_eval'

if os.path.exists(processed_test_images_dir):
    shutil.rmtree(processed_test_images_dir)

os.makedirs(os.path.join(processed_test_images_dir, 'No_DR'), exist_ok=True)
os.makedirs(os.path.join(processed_test_images_dir, 'DR'), exist_ok=True)

print("Processing and copying new test set images...")
processed_image_paths = []
processed_labels = []

for idx, row in test_new_df.iterrows():
    img_id = row['id_code']
    original_img_path = os.path.join(new_images_dir, img_id)
    
    if not os.path.exists(original_img_path):
        continue
    
    # هنا نعالج الصورة بحجم كبير أولاً (300x300)
    processed_img = preprocess_aptos_image_for_eval(original_img_path, target_size=(300, 300))
    
    if processed_img is not None:
        label = row['binary_type']  # 'No_DR' أو 'DR'
        output_folder = os.path.join(processed_test_images_dir, label)
        output_path = os.path.join(output_folder, img_id)
        cv2.imwrite(output_path, processed_img)
        
        processed_image_paths.append(output_path)
        processed_labels.append(label)

print(f"Processed {len(processed_image_paths)} images from the new test set.")

processed_test_df_eval = pd.DataFrame({
    'filepath': processed_image_paths,
    'class': processed_labels
})

print("Dataframe for processed test images prepared.")

# 3. دالة إنشاء batch generator لكل نموذج مع preprocessing و target size المناسب
def create_test_generator(processed_test_df, preprocess_func, target_size, batch_size=32):
    test_gen = ImageDataGenerator(preprocessing_function=preprocess_func)
    test_batches = test_gen.flow_from_dataframe(
        dataframe=processed_test_df,
        x_col='filepath',
        y_col='class',
        target_size=target_size,
        batch_size=batch_size,
        class_mode='binary',
        shuffle=False
    )
    return test_batches

# 4. إنشاء batch generators للثلاثة نماذج مع الحجم والمعالجة الصحيحة
test_batches_resnet = create_test_generator(processed_test_df_eval, resnet_preprocess_input, (224, 224))
test_batches_densenet = create_test_generator(processed_test_df_eval, densenet_preprocess_input, (224, 224))
test_batches_efficientnet = create_test_generator(processed_test_df_eval, efficientnet_preprocess_input, (300, 300))
```

### ***Find the best Thresholds***

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report

# افترض models ديكت فيها {اسم النموذج: (مسار الملف، test_batches)}
models = {
    'ResNet50V2': ('ResNet50V2.model', test_batches_resnet),
    'DenseNet201': ('DenseNet201.model', test_batches_densenet),
    'EfficientNetB3': ('EfficientNetB3.model', test_batches_efficientnet)
}

thresholds = [0.35, 0.65]

for model_name, (model_path, test_batches) in models.items():
    print(f"\n================= Evaluating {model_name} =================")
    
    # تحميل النموذج
    model = tf.keras.models.load_model(model_path, compile=True)
    
    # إعادة تعيين الـ generator للحصول على نتائج صحيحة
    test_batches.reset()
    
    # تنبؤات احتمالية
    predictions = model.predict(test_batches).ravel()
    
    y_true = test_batches.classes
    class_names = list(test_batches.class_indices.keys())
    
    # إنشاء مخطط المصفوفات الالتباسية لكل threshold
    fig, axes = plt.subplots(1, len(thresholds), figsize=(6 * len(thresholds), 5))
    if len(thresholds) == 1:
        axes = [axes]
    
    for i, thresh in enumerate(thresholds):
        y_pred_binary = (predictions > thresh).astype(int)
        
        cm = confusion_matrix(y_true, y_pred_binary)
        
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                    xticklabels=class_names, yticklabels=class_names,
                    ax=axes[i])
        axes[i].set_title(f'{model_name} - Threshold {thresh}')
        axes[i].set_xlabel('Predicted Label')
        axes[i].set_ylabel('True Label')
        
        print(f"\n--- Classification Report for {model_name} at Threshold {thresh} ---")
        print(classification_report(y_true, y_pred_binary, target_names=class_names))
    
    plt.tight_layout()
    plt.show()
```

> The threshold of 0.65 is better overall as it balances precision and recall across both classes, yielding higher accuracy and more reliable predictions for both DR and No_DR.
